```python
# ASSIGNMENT 04:
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.__balance = balance
        self.history = []

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Invalid balance!")

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self.history.append(f"Deposited {amount}")
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance!")
        elif amount > 0:
            self.__balance -= amount
            self.history.append(f"Withdrew {amount}")
        else:
            print("Invalid amount!")

    def check_balance(self):
        print(f"{self.account_holder}'s Balance: {self.__balance}")

    def transfer(self, other_account, amount):
        if amount > self.__balance:
            print("Transfer failed: Insufficient balance!")
        elif amount > 0:
            self.__balance -= amount
            other_account.__balance += amount

            self.history.append(f"Transferred {amount} to {other_account.account_holder}")
            other_account.history.append(f"Received {amount} from {self.account_holder}")
        else:
            print("Invalid transfer amount!")

    def show_history(self):
        print(f"\nTransaction History of {self.account_holder}:")
        for t in self.history:
            print(t)


acc1 = BankAccount("Haya", 1000)
acc2 = BankAccount("Ali", 500)

acc1.deposit(200)
acc1.withdraw(150)
acc1.transfer(acc2, 300)

acc1.check_balance()
acc2.check_balance()

acc1.show_history()
acc2.show_history()
```

```python
# ASSIGNMENT 05
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_details(self):
        print("Name:", self.name, "Age:", self.age)


class Student(Person):
    def __init__(self, name, age, gpa):
        super().__init__(name, age)
        self.gpa = gpa

    def get_details(self):
        print("Student:", self.name, "Age:", self.age, "GPA:", self.gpa)


class Teacher(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary

    def get_details(self):
        print("Teacher:", self.name, "Age:", self.age, "Salary:", self.salary)


s1 = Student("Haya", 18, 3.8)
t1 = Teacher("Ali", 40, 50000)

people = [s1, t1]

for person in people:
    person.get_details()
```

```python
# ASSIGNMENT 06
from abc import ABC, abstractmethod
import math


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


shapes = []

n = int(input("How many shapes do you want to create? "))

for i in range(n):

    choice = input("Enter shape (circle/rectangle/triangle): ")

    if choice == "circle":
        r = float(input("Enter radius: "))
        shapes.append(Circle(r))

    elif choice == "rectangle":
        l = float(input("Enter length: "))
        w = float(input("Enter width: "))
        shapes.append(Rectangle(l, w))

    elif choice == "triangle":
        b = float(input("Enter base: "))
        h = float(input("Enter height: "))
        shapes.append(Triangle(b, h))

total_area = 0

for shape in shapes:
    total_area += shape.area()

print("Total Area:", total_area)
```
